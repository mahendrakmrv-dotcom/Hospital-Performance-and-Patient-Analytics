# Filling null values with median for wait_time_min column

```python
df['wait_time_min'].isnull().sum()
```

```python
print(df['wait_time_min'].dtype)
```

### As the "wait_time_min" column is in object datatype, im converting that into 'int' datatype to perform fillna() to replace the NaN values with median

```python
df['wait_time_min'] = pd.to_numeric(df['wait_time_min'], errors='coerce').fillna(0).astype(np.int64)
```

```python
print(df['wait_time_min'].dtype)
```

```python
df['wait_time_min'].isnull().sum()
```

```python
df['wait_time_min'] = df['wait_time_min'].replace(0, np.nan)
```

```python
df['wait_time_min'] = df['wait_time_min'].fillna(df['wait_time_min'].mean())
```

```python
df['wait_time_min'] = df['wait_time_min'].round(2)
```

### Checking null values in hospital_stay_days, if so replacing it with mean values

### Column re-admission has mixture of boolean values, so im changing the values as only to 'Yes' or 'No'

```python
df['readmission_30d'].value_counts()
```

```python
def fix(x):
    if x in ['1','Yes','TRUE']:
        return 'Yes'
    else:
        return 'No'
```

```python
df['readmission_30d'] = df['readmission_30d'].apply(fix)
```

```python
df['readmission_30d'].value_counts()
```

```python
df['readmission_30d'].isnull().sum()
```
